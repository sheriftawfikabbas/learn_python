# Introduction to Programming in Python

This course is aimed at providing basic Python programming skills to students who have never programmed before. We will go slowly and in thorough detail, to make sure the ideas come through smoothly.

Each **Class** notebook is a lesson, and each **Lab** notebook is the exercises that go with it. Work through them in order — every lab assumes only what has been taught so far. You can run the notebooks online with Google Colab (no installation needed), or locally with Jupyter.

# Learning outcomes

On successful completion of this course, students will be able to:

**Knowledge:**
1. Explain the fundamentals of procedural programming.
2. Explain the principles of inheritance, composition and polymorphism.

**Skills:**
1. Design, develop and debug procedural, modular and object-oriented programs.
2. Apply inheritance and code re-use.
3. Create and use libraries and packages.
