## 5.1 Python scripts

So far we've been programming Python on Google Colab. It's pretty easy to code in Colab, because Colab was designed as a place for people to demonstrate and teach Python. However, to write Python *software*, we have to start learning how to write Python script *files*.

A Python script is just a text file with the extension `.py` that contains Python code. On your own computer, create a file called `hello.py` (using any text editor — Notepad on Windows, or TextEdit on Mac) and type this single line into it:

```
print('Hello! I am a Python script.')
```

Save it, open a terminal (or command prompt) in the same folder, and run it by typing:

```
python hello.py
```

(On Linux or Mac you may need to type `python3 hello.py` instead.) The computer executes all the statements in the file, one after another — and doesn't need Colab or the internet for it.

## 5.2 Creating and importing modules

We learned, when using functions, that we can re-use our code instead of copy-pasting it. But what if you want to re-use a *set* of functions, not just one function? And what if you also want to re-use variables?

Python supports a cool technique for code reusability: *modules*. A module is quite simple: it's just a Python code file. It has to have the Python file extension, `.py`, and can contain as many functions and variables as you wish.

So the script `hello.py` above is also a module! And here in this folder we already have a module called `abc_module.py`. It contains exactly this code:

```
a = 1
pi = 3.14159265

def addition(a, b):
    return a + b
```

To *use* a module from your code, type the `import` keyword followed by the module name — without the `.py` extension — like this:

````python
import abc_module

print(abc_module.a)
````

That's how we can access the contents of a module after importing it: we use the name of the module, followed by the `.` notation.

There are three different ways to import a module:
1. Using `import`, as above.
2. Using `import module_name as alias`.
3. Using `from module_name import something`. To import *everything* from `module_name`, we type `from module_name import *`.

In (2), we give the module an alias, or a nickname. For example:

````python
import abc_module as abc

print(abc.a)
````

As for (3), we would use this if we only need some of the things in the module, not the whole module:

````python
from abc_module import pi, addition

print(pi)
print(addition(3, 4))
````

Note that when we apply method (3), we don't need to use the module name, or alias, to access the variable — `pi` and `addition` are now directly available.

### The `dir()` function

To find out the names of the variables and functions defined in a module, let's say `abc_module`, you can use the function `dir(abc_module)`. The output also contains special system names that start with an underscore, `_` — just ignore those:

````python
import abc_module

print(dir(abc_module))
````

Python itself comes with many ready-made modules — the *standard library*. For example, the `math` module gives you mathematical functions and constants:

````python
import math

print(math.sqrt(2))
print(math.pi)
````

Other popular standard library modules are `random` (random numbers), `os` and `sys` (talking to your operating system), and `json` (we'll use it shortly).

## 5.3 File operations

### Creating a new file

To create a new file in Python we use the `open()` function:

```
f = open('filename', 'w')
```

The first argument is the file name; the second is the *mode* of file access:
- `'w'`: open the file for writing, creates it if it does not exist (but empties it if it does!)
- `'a'`: open the file for appending, creates it if it does not exist
- `'r'`: open the file for reading; gives an **error** if it does not exist
- `'x'`: creates the file; gives an **error** if the file already exists

### Writing into a file

To write text into an opened file, we use the `write()` function:

```
f = open('filename', 'w')
f.write('hello!')
```

### Closing the file

When you are done and you don't need to access the file any longer in your program, you must close it with the `close()` function. If you don't, your text may never actually be saved to disk:

```
f = open('filename', 'w')
f.write('hello!')
f.close()
```

### Reading from a file

The `readlines()` function returns a list of all the lines in the file, one list item per line. Note that this time we open the file in **read** mode, `'r'`:

```
f = open('filename', 'r')
l = f.readlines()
for i in l:
    print(i)
f.close()
```

Let's do the full write-then-read cycle for real:

````python
# write two lines into a new file
f = open('demo.txt', 'w')
f.write('hello!\n')
f.write('this is the second line\n')
f.close()

# read the lines back and print them
f = open('demo.txt', 'r')
l = f.readlines()
for i in l:
    print(i)
f.close()

# tidy up: delete the demo file again
import os
os.remove('demo.txt')
````

## 5.4 JSON file input-output

In Class 9 we'll work a lot with JSON data, so let's see how to store it in a file. The `json` module can *dump* a Python collection (like a dictionary or a list) into a file, and *load* it back later:

````python
import json

xo = [[1, 2, 3]]

# save the collection into a file
f = open('myfile.json', 'w')
json.dump(xo, f)
f.close()

# ... and read it back
f = open('myfile.json', 'r')
xo = json.load(f)
print(xo)
f.close()

os.remove('myfile.json')
````

## Summary

- A Python *script* is a `.py` text file; run it from a terminal with `python hello.py`.
- A *module* is any `.py` file you can `import`; access its contents via `module.name`, via an alias (`import m as x`), or import names directly (`from m import x`).
- `dir(module)` lists the names defined in a module; the standard library includes `math`, `random`, `os`, `sys` and `json`.
- Files are opened with `open(name, mode)` where the mode is `'w'` (write), `'a'` (append), `'r'` (read) or `'x'` (create-new); always `close()` when done.
- `json.dump()` saves a collection to a JSON file; `json.load()` reads it back.
