## 10.1 Scraping the internet using the `BeautifulSoup` library

Scraping a website means: visiting pages on the website and downloading a copy of those pages. Typically, the scraping algorithm will navigate the website by visiting the links on each page, and then visiting and downloading the code in the linked pages.

But a downloaded page is a wall of HTML text. Couldn't we just use string methods, like `replace()`, to pull information out of it? Let's try: we want to extract the text inside an `<h1>` heading.

```python
a = "<h1 class='some class'>I need to extract this text.</h1>"

# try to remove the opening tag with a simple replace:
b = a.replace('<h1>', '')
print(b)
```

Notice the problem: the opening tag was `<h1 class='some class'>`, so our exact string `'<h1>'` never matched, and nothing was removed (only the exact closing tag `</h1>` was matched and removed). HTML tags come in endless variations — string hacks just can't keep up. We need a real *HTML parser*: `BeautifulSoup`.

### BeautifulSoup on a string of HTML

Let's parse a small HTML string — no internet needed:

```python
from bs4 import BeautifulSoup

html = '''<html><body>
<h1>Three languages</h1>
<ul>
  <li><a href="https://www.python.org">Python</a></li>
  <li><a href="https://www.r-project.org">R</a></li>
  <li>Scratch</li>
</ul>
</body></html>'''

data_bs = BeautifulSoup(html, 'html.parser')

print('The heading text:', data_bs.find('h1').text)
print('The first link:', data_bs.find('a'))

list_of_links = data_bs.find_all('a')
print('Number of links:', len(list_of_links))
for link in list_of_links:
    print(link.text, '->', link.get('href'))
```

Now the real thing. The cell below downloads a live page (you need internet access for this and the next cells). The page is *huge*, so we print only the first 800 characters:

```python
import requests

response = requests.get("https://www.mathpractice.xyz")
data = response.text
print(data[:800], '\n...')
```

```python
from bs4 import BeautifulSoup
import requests

response = requests.get("https://abc.com")
data = response.text

data_bs = BeautifulSoup(data, 'html.parser')

print("First link using find():")
print(data_bs.find('a'))

list_of_links = data_bs.find_all('a')
print('Number of links on the page:', len(list_of_links))

print('The first 10 link addresses:')
for link in list_of_links[:10]:
    print(link.get('href'))
```

Several things happened in these cells:
- First of all, to use `BeautifulSoup`, we import it from the `bs4` package and create a `BeautifulSoup` object from the page text (the second argument `'html.parser'` tells it which parser to use).
- Note that the `response` object gives access to the web page content via the `text` attribute. (You can also convert JSON content into a dictionary using the `json()` method, as we did in Class 9.)
- You can get a list of all HTML tags of one kind using the `find_all()` method of `BeautifulSoup`.
- In each link tag, you can obtain the content of the `href` attribute using the `get()` method.

## Features of the `BeautifulSoup` library

You can dig into web pages using a large range of methods and properties that are provided by `BeautifulSoup`, and which we will be using in the lab:

- `find(name)`: finds the **first** HTML tag with the given name
- `find_all(name)`: finds **all** HTML tags with the given name, as a list
- `.text` (property): the text inside a tag
- `get(attr)`: gets a particular attribute value from a tag
- `attrs` (property): the attributes of a tag, as a dictionary
- `findChildren()`: the children of a tag

## Summary

- Scraping = downloading web pages and extracting information from their HTML.
- String methods like `replace()` are too fragile for HTML — use a parser.
- `requests.get(url).text` downloads the page; `BeautifulSoup(text, 'html.parser')` parses it.
