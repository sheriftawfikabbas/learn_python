## 1.1 A bit about Python
Python is quite a nice and easy language to learn and understand. In fact, it is one of the easiest languages to hit the road running with. This is because you can virtually run your Python code, or script, anywhere: on your laptop or even on your phone. This is thanks to the many online Python servers that will allow you to run Python code online. 

Unlike languages such as Java or C — where you must first *compile* your code into an executable file before it can run — Python is an *interpreted* language: a program called the Python interpreter reads your code and executes it directly, one statement at a time. That's part of why Python feels so immediate to use.

Let me give you an example of how simple it is to run Python code.

## 1.2 The print function
Let's start with the simplest thing you can do in a program: to ask the computer to print out something, such as `"Like coffee?"`

How do we get the computer to print out this question on the screen? For that, we use the `print()` function. This function orders the computer to print *something* to the screen. That something is called a *string*, which we will deal with shortly.

Let's write that code, which will be our first ever line of Python code in this course:

```python
print('Like coffee?')
```

Do you notice something in the above line? It looks different from normal text on this page. It is actually an *executable* line of code! This means that, when you press the play button to the left of the print statement, the program will actually **run**! That's without needing to install any software on any computer. This is the power of using Google's Colab: it lets you execute Python code on the fly.

So what just happened? We wrote a print statement, in which we called the `print` function. It is a function because it receives something, the `"Like coffee?"` text, and then performs some task with the text.

Let's try something else. Let's print two sentences:

```python
print("Like coffee?",'"Yes, with milk please!"')
```

Did you notice what happened here? The print function received two, not one, strings. They are separated by a comma. The print function then prints both sentences one after the other, leaving a space in between.

Now let's try to understand how we write strings like `"Like coffee?"`

## 1.3 Strings

A string is any collection of characters, numbers, dots, symbols, or anything else you can access via the keyboard (or beyond!), enclosed between two single quotes, or between two double quotes. So, for example, typing `"a"` is a valid string in Python:

```python
"a"
```

Here we just typed the string `"a"`, and Python only repeated the value of that string, which again is `'a'`. Did you spot the difference?

A string in Python is composed of three things: an opening quotation mark, a set of characters, and a closing quotation mark. Python accepts two types of quotation marks: single and double. So we can also type `'a'` up there, and we will get the same result.

The string is an example of a Python *data type*: it is something that represents a value.

## 1.4 Python data types: strings, numbers and boolean values

Python has a number of data types that enable you to perform various types of operations on data. Here we discuss the fundamental three data types of Python: strings, which we just discussed above, numbers and logical, or boolean, values.

Numbers are just numbers! Type a number in the Python interpreter, say `5`, and you will just get `5` back.

```python
5
```

Well you can actually do arithmetic right there, in the Python interpreter.

```python
5+6-8*3
```

Here I just did a complex arithmetic operation using integers, and Python gave me the result of the operations. What I typed up here is called, in programming terminology, an *expression*: an operation on values that yields a value. Here the value we got is `-13`, which is the value of the expression.

Numbers in Python can be integers, decimal numbers with a few decimal places, decimals with a very large number of decimal places, and very large numbers.

For strings, when you typed `"a"` in the Python interpreter, that also was an expression: a string expression. A string can be empty, like `""`, or can have as many characters as you want.

Another important data type is the boolean data type. This one has only two possible values: `True` and `False`. Let's try that.

```python
True
```

Those two data values represent the two possible outcomes of a yes/no question, such as: is that ball blue? Soon, we will find out that this type of question is very common in programming.

## 1.5 Variables

A variable in Python, as well as in any programming language, enables you to store a value in memory. As an example, let us create a variable that will store the string value "I love coffee" in memory.

```python
s = "I love coffee"
```

We have just created a variable that has two important features: a name, which is `s`, and a value stored inside it, which is `"I love coffee"`. This memory storage was performed via the `=` sign. The entire line, `s = "I love coffee"` is called an *assignment statement*, and the `=` is called the *assignment operator*.

So that was our first operator here, the `=`. This operator is unique in that it is different from the arithmetic operators we know, such as the `+` and `-`, because `=` does not give you an expression. It is rather a statement. A statement orders the computer to perform a task, such as the print statement, while an expression is only meant to return a value.

**Remark:** Note the difference between expressions and statements.

## 1.7 Arithmetic operations: `+`, `-`, `*`, `/`, `%`, `//`

Back to operators. You can work with the standard maths operators in Python, which are `+`, `-`, `*`, `/`, `//` and `%`. The first four are familiar, but `//` and `%` might need an introduction.

The `//` operator is related to `/`: it gives you the *integer quotient* of the division, and therefore it is called the floor-division operator. That is, it just drops the decimal part of the division result. For example, `9 // 4` gives `2`.

`%` is the modulus operator. It is related to `//` in that: while `a // b` gives you the integer *quotient*, `%` gives the *remainder*. For example, in the division `9 / 4 = 2 + 1/4`, the quotient is `2` and the remainder is `1`. So in Python, `9 % 4` gives `1`.

```python
print(9 / 4)   # ordinary division
print(9 // 4)  # the integer quotient (floor division)
print(9 % 4)   # the remainder (modulus)
```

## 1.8 Comparison operations: `==`, `!=`, `<`, `>`, `<=`, `>=`

Now that we know how to use operators on numbers to create numbers, the operators below create boolean values. This is because they ask questions about the variables they operate on.

- `a == b` means: is `a` exactly equal to `b`?
- `a != b` means: is `a` **not** equal to `b`?
- `a < b` means: is `a` less than `b`?
- `a > b` means: is `a` greater than `b`?
- `a <= b` means: is `a` less than **or equal to** `b`?
- `a >= b` means: is `a` greater than **or equal to** `b`?

Let's evaluate some expressions based on these operators:

```python
#This code demonstrates operations
a = 3
b = 5
c = a == b
print(c)
d = a > b
print(d)
e = a <= b
print(e)
```

## 1.9 Operator precedence and brackets

Remember BODMAS from high school? It stands for: Brackets, Of, Division, Multiplication, Addition and Subtraction. It also applies in Python the same way it applied in your maths lessons. For example, `2 + 3 * 4` evaluates to `14`, while `(2 + 3) * 4` evaluates to `20`.

## 1.12 More on data types, and getting and setting a data type

In Python, there are three numerical data types, not just one: `int`, `float`, and `complex`. The `int` is an integer, which means the whole numbers, both positive and negative. Examples: `7`, `1023`, `-2982`, etc. A `float` is a positive or a negative decimal number, such as `4.2`, `-50.298`. Let's worry about `complex` later in this course.

### Retrieving the data type of a variable

If you have some variable, Python lets you find out its data type using the function `type()`.

```python
b=45
print(type(b))
```

Here, `class 'int'` tells you that `b` is an integer.

### Setting the data type of a variable

In Python, you set the data type of a variable by one of two methods:
1. By assigning a value to it
2. By using a data type constructor function

So when you type `a = 4`, you have set the data type of variable `a` to an `int` data type.

**Remark:** The freedom of setting a variable type in Python makes it fundamentally different from other languages such as Java, C and C#. These languages apply *strong typing* on their variables, which means that you must specify the type of a variable on creation, and cannot change the type of the variable later in the code.

To set the data type using a constructor function, there is a constructor function for each data type that we can choose from. They are:
- For integers: `int()`
- For floats: `float()`
- For strings: `str()`

Before we use a constructor function, we need to think about the purpose of using it to change the data type of a variable. If you have a float, why would you want to turn it into an integer? A simple reason: to get rid of the decimal part! (Note that `int()` *truncates* the number — it does not round it.)

```python
a = 54.67
b = int(a)
print(b)
```

### Setting multiple variables

In Python, you can set values into several variables in the same statement. For example, to set the value of `a` to `1` and `b` to `2`, we do the following:

```python
a,b=1,2
```

## 1.13 Python statement termination

In Python, as you have noticed already, a *statement* starts at the start of the line and ends with a new line. So you cannot put two statements in the same line — typing the following would give you an error: `a=3 print(a)`.

There is another way you can terminate a statement: by using the semicolon, `;`, just like in Java, C, C# and other languages. Although it is optional, you will need to use it if you want to put two statements on the same line e.g. `a=3; print(a)`.
